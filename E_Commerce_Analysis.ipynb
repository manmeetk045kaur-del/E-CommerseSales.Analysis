### 1. Loading the Dataset and Inspecting Structure

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")

# Display first 5 rows
print("First 5 Rows:")
display(df.head())

# Check data types and structure
print("\nDataset Info:")
df.info()
```

### 2. Overall Sales Statistics and Aggregations

```python
# Calculate key metrics across the dataset
total_sales = df['Total_Sales'].sum()
avg_sales = df['Total_Sales'].mean()
max_sale = df['Total_Sales'].max()
min_sale = df['Total_Sales'].min()
total_qty = df['Quantity'].sum()
total_orders = df['Order_ID'].count()

print(f"Total Revenue (Sales): ₹{total_sales:,.2f}")
print(f"Average Order Value: ₹{avg_sales:,.2f}")
print(f"Highest Single Order Sale: ₹{max_sale:,.2f}")
print(f"Lowest Single Order Sale: ₹{min_sale:,.2f}")
print(f"Total Quantity Sold: {total_qty}")
print(f"Total Number of Orders: {total_orders}")
```

### 3. Groupby Analysis by Category

```python
# Group by product category to check total sales, average order value, and order count
category_analysis = df.groupby('Category')['Total_Sales'].agg(
    Total_Revenue='sum',
    Average_Order_Value='mean',
    Order_Count='count'
).sort_values(by='Total_Revenue', ascending=False)

print("Performance by Category:")
print(category_analysis)
```

### 4. Groupby Analysis by City and Region

```python
# Analyze sales performance across different cities
city_analysis = df.groupby(['City', 'Region'])['Total_Sales'].agg(
    Total_Revenue='sum',
    Order_Count='count'
).reset_index().sort_values(by='Total_Revenue', ascending=False)

print("Top Performing Cities:")
print(city_analysis)
```

### 5. Payment Method & Top Products Analysis

```python
# Analyze total sales by payment method
payment_analysis = df.groupby('Payment_Method')['Total_Sales'].agg(
    Total_Revenue='sum',
    Order_Count='count'
).sort_values(by='Total_Revenue', ascending=False)

print("Sales by Payment Method:")
print(payment_analysis)

# Identify top 5 best-selling products by total revenue
top_products = df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False).head(5)
print("\nTop 5 Products by Revenue:")
print(top_products)
```

### 6. Filtering and Sorting Operations

```python
# Filter orders where Total_Sales is greater than ₹10,000 and rating is above 4.0
high_value_orders = df[(df['Total_Sales'] > 10000) & (df['Rating'] > 4.0)]

print(f"Number of High-Value & Highly Rated Orders: {len(high_value_orders)}")
print(high_value_orders[['Order_ID', 'Customer', 'Category', 'Product', 'Total_Sales', 'Rating']])
```
